### 🔍 **Scenario: Understanding the Impact of Content Duration on Netflix's Top 10 Lists**

This scenario helps you develop an end-to-end project for your portfolio.

Background: As a data scientist at Netflix, you're tasked with exploring the dataset containing weekly top 10 lists of the most-watched TV shows and films. For example, you're tasked to find out what the relationship is between duration and ranking over time. Answering this question can inform content creators and strategists on how to optimize their offerings for the platform.

**Objective**: Determine if there's a correlation between content duration and its likelihood of making it to the top 10 lists.

You can query the pre-loaded CSV files using SQL directly. Here’s a **sample query**:

```python
SELECT show_title, MAX(cumulative_weeks_in_top_10) as max_cumulative_weeks_in_top_10
FROM 'netflix_top10_country.csv'
WHERE country_name = 'Argentina'
GROUP BY show_title
ORDER BY max_cumulative_weeks_in_top_10 DESC
LIMIT 3;
```

```python
import pandas as pd

# 0. 先讀取全球榜的 CSV 檔案
global_top_10 = pd.read_csv("netflix_top10.csv")

# 1. 確保週別與觀看時數的資料型態正確
global_top_10['week'] = pd.to_datetime(global_top_10['week'])
global_top_10['weekly_hours_viewed'] = pd.to_numeric(global_top_10['weekly_hours_viewed'], errors='coerce')

# 2. 依照「每週(week)」進行分組，並在每週內依據「觀看時數」由高到低排序
weekly_combined = global_top_10.sort_values(by=['week', 'weekly_hours_viewed'], ascending=[True, False])

# 3. 重新計算每週不分類別的總排名（取前10名）
weekly_all_categories_top10 = weekly_combined.groupby('week').head(10).copy()

# 4. 建立一個新的欄位顯示總榜排名
weekly_all_categories_top10['overall_weekly_rank'] = weekly_all_categories_top10.groupby('week').cumcount() + 1

# 查看結果
weekly_all_categories_top10[['week', 'overall_weekly_rank', 'show_title', 'category', 'weekly_hours_viewed']].head(10)
```

```python
import pandas as pd

# 1. 讀取各國榜資料（拿掉 index_col=0，確保 country_name 欄位在裡面）
countries_top_10 = pd.read_csv("netflix_top10_country.csv")

# 2. 篩選出台灣（Taiwan）的資料
taiwan_top_10 = countries_top_10[countries_top_10['country_name'] == 'Taiwan'].copy()

# 3. 確保週別資料型態正確
taiwan_top_10['week'] = pd.to_datetime(taiwan_top_10['week'])

# 4. 計算在台灣榜上，「累積在榜週數」最高的前 10 名
taiwan_longest_charts = taiwan_top_10.groupby(['show_title', 'category'])['cumulative_weeks_in_top_10'].max().reset_index()
taiwan_longest_charts = taiwan_longest_charts.sort_values(by='cumulative_weeks_in_top_10', ascending=False)

# 查看台灣霸榜前 10 名結果
taiwan_longest_charts.head(10)
```
